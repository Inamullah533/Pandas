```python
import numpy as np
import pandas as pd
```

```python
Data_Employee = pd.DataFrame ({
    'Employee_Id' : [1,2,3,4,5,6],
    'Employee_Name' : ['Ali','Ahmad','Zain','Kashif','Hassan','Sahil'],
    'Department' : [ 'HR ','Employee','Employee','HR','IT','Finance']
})

Salaries = pd.DataFrame ({
    'Employee_Id': [1,2,3,4,7,8],
    'Salary' : [50000,35000,38000,55000,40000,37000],
     'Bonus' : [20000,10000,12000,23000,17000,15000]
        })
```

```python
Data_Employee
```

```python
Salaries
```

```python
pd.merge(Data_Employee,Salaries)
```

```python
pd.merge(Data_Employee,Salaries,how ='outer')
```

```python
pd.merge(Data_Employee,Salaries,how ='inner')
```

```python
pd.merge(Data_Employee,Salaries,how ='right')
```

```python
pd.merge(Data_Employee,Salaries,how ='left')
```

```python
df1 = pd.DataFrame ({
    'A' : ['A0','A1','A2'],
    'B ': ['B0','B1','B2'],
    'C' : ['C0','C1','C2']
})
df2 =pd.DataFrame ({
    'A' : ['A3','A4','A5'],
    'B' : ['B3','B4','B5'],
    'C' : ['C3','C4','C5']
})
    
```

```python
df1
```

```python
df2
```

```python
pd.concat([df1,df2])
```

```python
pd.concat([df1,df2],axis = 1)
```
