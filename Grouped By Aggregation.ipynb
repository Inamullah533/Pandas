```python
import numpy as np
import pandas as pd
```

```python
Data = {
    'Categary' : ['A','B','A','B','A','B','A','B'],
    'Stores' : ['S1','S2','S2','S1','S1','S2','S1','S2'],
    'Sales' : [100,450,260,490,156,260,240,350],
    'Quantity' : [10,12,8,20,12,15,11,5],
    'Date' : pd.date_range('12-Oct,2026',periods = 8)
}
df = pd.DataFrame(Data)
```

```python
df
```

```python
cat = df.groupby('Categary')
cat
for i,v in cat:
    print(i)
    print(v)
```

```python
cat = df.groupby('Categary')['Sales'].sum()
cat
```

```python
cat = df.groupby('Stores')
cat
for i , v in cat:
    print (i)
    print (v)
```

```python
cat = df.groupby('Stores')['Sales'].sum()
cat
```

```python
##GROUPBY MULTIPLE:
##GROUPBY CATEGORY AND STORE:

cat = df.groupby(['Categary','Stores'])
cat
for i , v in cat:
    print(i)
    print(v)
```

```python
##SUM:
cat = df.groupby(['Categary','Stores'])['Sales'].sum()
cat
```
